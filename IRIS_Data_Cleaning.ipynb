### **BASIC DATA CLEANING**

```python
## IMPORT LIBRARIES
import pandas as pd 
import numpy as np

## LOAD THE DATASET
df = pd.read_csv('IRIS_Dataset.csv')  

# Copy the DataFrame to avoid modifying the original dataset
data = df.copy()
data
```

#####  ✅ Remove Duplicates Records -- Delete duplicate rows.

```python
data.duplicated() # Check for duplicate rows
```

```python
# Before removing duplicates, let's check how many duplicate rows are present in the dataset
data.duplicated().sum() # Count the number of duplicate rows
```

```python
new_data = data.drop_duplicates() # Remove duplicate rows and return a new DataFrame without duplicates
```

```python
# After removing duplicates, let's check how many duplicate rows are present in the new dataset
new_data.duplicated().sum() 
```

##### ✅ Fixing Inconsistent Formats -- Standardize the format of data in columns.

```python
# Changing the values of the 'Species' column to more readable format

# data.loc[['row_condition'], ['column_name']] = new_value , 
# .loc is used to access a group of rows and columns by labels or a boolean array.
mask1 = new_data["species"].eq("Iris-setosa") # Create a boolean mask for rows where 'Species' is 'Iris-setosa'
new_data.loc[mask1, "species"] = "Setosa"
mask2 = new_data["species"].eq("Iris-versicolor") # Create a boolean mask for rows where 'Species' is 'Iris-versicolor'
new_data.loc[mask2, "species"] = "Versicolor"
mask3 = new_data["species"].eq("Iris-virginica") # Create a boolean mask for rows where 'Species' is 'Iris-virginica'
new_data.loc[mask3, "species"] = "Virginica"

new_data  # Display the cleaned dataset with updated 'Species' values
```

##### ✅ Fixing Impossible values -- Correct or Replace the impossible values with NaN.
###### **NOTE** :  NaN will further be handled in the Data Preprocessing step 

```python
# Any type of length can't be negative. So, we will replace any negative values in the dataset with NaN (Not a Number).

new_data.loc[new_data['sepal_length'] < 0, 'sepal_length'] = np.nan  # Replace negative values in 'sepal_length' with NaN
new_data.loc[new_data['sepal_width'] < 0, 'sepal_width'] = np.nan  # Replace negative values in 'sepal_width' with NaN
new_data.loc[new_data['petal_length'] < 0, 'petal_length'] = np.nan  # Replace negative values in 'petal_length' with NaN
new_data.loc[new_data['petal_width'] < 0, 'petal_width'] = np.nan  # Replace negative values in 'petal_width' with NaN

new_data # Display the dataset after replacing negative values with NaN
```
